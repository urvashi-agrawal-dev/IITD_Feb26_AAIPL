## 📌 Project Overview

This project was developed for the AMD AI Premier League (AAIPL).

We built two AI agents:

- 🤔 Question Agent (Q-Agent) – Generates logical reasoning and puzzle-based MCQs.
- 🤖 Answer Agent (A-Agent) – Solves and answers opponent questions.

The system strictly follows required JSON format and tournament constraints.


## 🧠 Topics Covered

1. Logical Reasoning – Syllogisms
2. Puzzles – Seating Arrangements
3. Blood Relations & Family Tree
4. Alphanumeric Series


## ⚙️ How To Run

### Generate Questions
```bash
python -m agents.question_agent --num_questions 20 --verbose
```

### Generate Answers
```bash
python -m agents.answer_agent --input_file outputs/questions.json --verbose
```

## 📊 Evaluation Logic

A-agent Score = (Correct Answers / N) × 100

Q-agent Score = (Incorrectly answered by opponent / N) × 100

Agents must pass format checks and time limits.
